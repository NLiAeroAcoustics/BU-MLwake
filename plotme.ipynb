```python
from scipy.io import savemat, loadmat
import matplotlib.pyplot as plt
import matplotlib
```

```python
idx = 222
field = 'vx_test'
test_dat = loadmat('Results/test_prediction.mat')
plt.figure(figsize=(12, 6), dpi=60)
#print(test_dat['label'][idx])
# 1st plot
plt.subplot(1, 2, 1)
pic = test_dat[field][idx, :, :]
my_cmap = matplotlib.colors.LinearSegmentedColormap.from_list("", ["blue","lime","yellow"])
img_re_plot = plt.imshow(pic.T,
                         vmin = 0.65*test_dat[field][idx, :, :].max(), 
                         vmax = test_dat[field][idx, :, :].max(),
                         cmap = my_cmap)
plt.gca().invert_yaxis()
plt.colorbar(fraction=0.025, pad=0.02)
plt.title('CFD', fontsize=15)

# 2nd plot
plt.subplot(1, 2, 2)
pic = test_dat[field+'_pred'][idx, :, :]
my_cmap = matplotlib.colors.LinearSegmentedColormap.from_list("", ["blue","lime","yellow"])
img_re_plot = plt.imshow(pic.T,
                         vmin = 0.65*test_dat[field][idx, :, :].max(), 
                         vmax = test_dat[field][idx, :, :].max(),
                         cmap = my_cmap)
plt.gca().invert_yaxis()
plt.colorbar(fraction=0.025, pad=0.02)
plt.title('ML predicted', fontsize=15)
```
